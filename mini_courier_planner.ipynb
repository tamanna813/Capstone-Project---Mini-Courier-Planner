# Mini-Courier Planner
Run the cells in order. This notebook implements sorting, 0/1 knapsack, Dijkstra, Nearest-Neighbour TSP and runtime benchmarking.

```python
from pathlib import Path
import sys
PROJECT = Path.cwd()
if not (PROJECT / 'datasets').exists() and (PROJECT.parent / 'datasets').exists(): PROJECT = PROJECT.parent
sys.path.insert(0, str(PROJECT / 'src'))
from main import *
```

```python
parcels = load_parcels(PROJECT / 'datasets' / 'parcels.csv')
graph = load_graph(PROJECT / 'datasets' / 'routes.csv')
ordered = prioritize_parcels(parcels)
print('Prioritized parcels:')
for p in ordered: print(p)
```

```python
selected, total_weight, total_value = knapsack_select(ordered, 15)
print('Selected parcels:', [p.parcel_id for p in selected])
print('Total weight:', total_weight, 'kg')
print('Total value:', total_value)
```

```python
shortest = all_pairs_shortest_paths(graph)
destinations = sorted({p.destination for p in selected})
route, route_distance = nearest_neighbour_tsp(shortest, 'Warehouse', destinations)
print('Route:', ' -> '.join(route))
print('Round-trip distance:', route_distance, 'km')
```
